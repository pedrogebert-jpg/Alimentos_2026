```python
import pandas as pd

# Carregar o arquivo CSV
df = pd.read_csv('/content/fisico-quimicas.csv')

# Selecionar as colunas desejadas
df_selected = df[['Amostra', 'Categoria', 'Acucar_g_100g', 'Acucar_estimado']]

# Exibir todas as linhas do DataFrame resultante
display(df_selected)
```

Pergunta 1) Os valores de açúcar estimado acompanham os valores de açúcar medidos nas amostras? Apresente um gráfico que permita avaliar essa relação.

```python
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, linregress

# Criar o gráfico de dispersão com a linha de regressão
plt.figure(figsize=(10, 6))
sns.regplot(x='Acucar_g_100g', y='Acucar_estimado', data=df_selected, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})

# Calcular a regressão linear simples
slope, intercept, r_value, p_value, std_err = linregress(df_selected['Acucar_g_100g'], df_selected['Acucar_estimado'])

# Calcular o coeficiente de correlação de Pearson
correlation, _ = pearsonr(df_selected['Acucar_g_100g'], df_selected['Acucar_estimado'])

# Adicionar título e rótulos
plt.title('Relação Açúcar 100g versus Açúcar estimado')
plt.xlabel('Acucar_g_100g')
plt.ylabel('Acucar_estimado')

# Adicionar equação da reta e coeficiente de correlação ao gráfico
plt.text(0.05, 0.95, f'Regressão Linear: y = {slope:.2f}x + {intercept:.2f}', transform=plt.gca().transAxes, fontsize=12, verticalalignment='top')
plt.text(0.05, 0.90, f'Correlação de Pearson: r = {correlation:.2f}', transform=plt.gca().transAxes, fontsize=12, verticalalignment='top')

plt.grid(True)
plt.show()
```

Resposta: Sim, os valores acompanham uns aos outros.
