```python
# Extraindo a terceira coluna (índice 2) do DataFrame df
coluna_3_dados = df.iloc[:, 2]

# Calcular a frequência absoluta
frequencia_absoluta_col3 = coluna_3_dados.value_counts()

# Calcular a frequência relativa
frequencia_relativa_col3 = coluna_3_dados.value_counts(normalize=True)

# Calcular a porcentagem
porcentagem_col3 = coluna_3_dados.value_counts(normalize=True) * 100

# Criar um DataFrame para exibir os resultados de forma organizada para a coluna 3
analise_grau_dulcor = pd.DataFrame({
    'Frequência Absoluta': frequencia_absoluta_col3,
    'Frequência Relativa': frequencia_relativa_col3.round(2),
    'Porcentagem (%)': porcentagem_col3.round(2)
})

display(analise_grau_dulcor)
```

Interpretação: A maioria das pessoas prefere um cereal doce, sem ser excessivo. A minoria, porém, prefere que o cereal seja nada doce

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Extraindo a sexta coluna (índice 5) do DataFrame df
coluna_6_dados = df.iloc[:, 5]

# Convertendo os dados para tipo numérico, se necessário, para garantir que o histograma funcione corretamente
# Usando .str.replace(',', '.') para lidar com vírgulas como separadores decimais, se presentes, e depois convertendo para float
# Usando errors='coerce' para converter valores não numéricos em NaN
coluna_6_dados_numerica = coluna_6_dados.astype(str).str.replace(',', '.').astype(float)

# Determinando o preço mínimo e máximo para definir as faixas
min_price = coluna_6_dados_numerica.min()
max_price = coluna_6_dados_numerica.max()

# Criando faixas de preço (bins) com um intervalo de 1 unidade, cobrindo todo o range
# Arredondamos para baixo o mínimo e para cima o máximo para garantir que todos os valores sejam incluídos
bins = np.arange(np.floor(min_price), np.ceil(max_price) + 1, 1)

# Criar o histograma com as faixas de preço personalizadas
fig = plt.figure(figsize=(12, 6))
sns.histplot(coluna_6_dados_numerica, bins=bins, kde=False, color='skyblue', edgecolor='black')
plt.title('Distribuição de Preços de Cereal')
plt.xlabel('Faixa de Preço (R$)')
plt.ylabel('Frequência')
plt.xticks(bins) # Define os ticks do eixo X nos limites das faixas
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
