**Variável aleatória:** Uma variável aleatória é uma variável que atribui um valor numérico aos possíveis resultados de um experimento aleatório. Como o resultado do experimento não é conhecido antecipadamente, o valor da variável pode mudar a cada repetição.

**Tipos de variáveis aleatórias:** 1) Discreta: Assume valores que podem ser contados um a um.

Exemplo de variáveis aleatórias discretas na indústria de alimentos:

Número de embalagens defeituosas em um lote de 1.000 latas de milho inspecionadas na linha de produção (valores inteiros: 0, 1, 2...)

Quantidade de ovos quebrados encontrados em uma caixa comercializada com 30 unidades na seção de hortifrúti.

2) Contínua: Assume qualquer valor dentro de um intervalo numérico contínuo (números reais). Entre dois valores quaisquer, sempre existirá outro valor possível.

Exemplo de variáveis aleatórias contínuas na indústria de alimentos:

Teor de umidade de um lote de grãos de café armazenados em silos (medido em porcentagem, podendo assumir valores como 11,4% , 14,6%, etc)

pH de um lote de iogurte natural medido durante o processo de fermentação na etapa de controle de qualidade. O pH do mosto ou da cerveja em maturação é medido continuamente (ex: 4,35; 4,42; 4,50)

**Modelo de Bernoulli:** O modelo de Bernoulli é um modelo estatístico usado quando temos um experimento que possui apenas dois resultados possíveis. Um dos resultados é denominado "probabilidade de sucesso" (1), e a outra, denominado "probabilidade de fracasso"(0).
 Por exemplo: uma máquina de embalagens tem 10% (p= 0.10) de chance para a embalagem sair defeituosa, se definirmos o defeituoso como sucesso, temos P(X=1) = 0.10 , ea chance da embalagem não sair defeituosa é P(X=0 )= 1-0.10= 0.9.

10% de sucesso (defeituoso), 90% de fracasso (não defeituoso)

**Modelo Binomial**: O modelo Binomial é usado para representar o número de sucessos obtidos em um número fixo de tentativas independentes, mantendo a probabilidade de sucesso em cada tentativa. Esse modelo é utilizado para situações onde a repetição de uma ação não muda a chances de um sucesso na próxima ação, como o lançamento de uma moeda, a probabilidade da moeda sempre será 50%, não importando quantas vezes uma pessoa lança essa moeda.

**Modelo de Poisson**: O modelo de Poisson é uma distribuição de probabilidade utilizada para representar o número de vezes que determinado evento ocorre em um intervalo fixo de tempo, espaço ou quantidade. Ele é aplicado principalmente quando estamos interessados em contar a ocorrência de eventos e existe uma taxa média de ocorrência conhecida.
Na área de alimentos, o modelo de Poisson pode ser utilizado, por exemplo, para analisar a quantidade de embalagens defeituosas encontradas em um lote de produtos. Se uma indústria encontra, em média, 3 embalagens defeituosas a cada 1.000 unidades inspecionadas, a distribuição de Poisson pode ser utilizada para calcular a probabilidade de encontrar uma determinada quantidade de embalagens defeituosas em um novo lote.

A fórmula da distribuição de Poisson é:

P(X = x) = (e⁻λ · λˣ) / x!  , onde:

X = número de ocorrências;
x = número específico de ocorrências que queremos analisar;
λ (lambda) = número médio de ocorrências no intervalo;
e ≈ 2,718;
x! = fatorial de x.

**Comparação entre os modelos:** Os modelos de Bernoulli, Binomial e Poisson são distribuições de probabilidade utilizadas para estudar variáveis aleatórias discretas, porém apresentam aplicações diferentes.

O modelo de Bernoulli é utilizado quando há apenas um experimento com dois resultados possíveis, como sucesso ou fracasso. Na área de alimentos, um exemplo seria verificar se uma embalagem está defeituosa ou não defeituosa.

O modelo Binomial é utilizado quando um experimento de Bernoulli é repetido várias vezes, com um número fixo de tentativas. Por exemplo, pode-se calcular a probabilidade de encontrar exatamente 5 embalagens defeituosas em uma amostra de 100 embalagens.

Já o modelo de Poisson é utilizado para determinar a probabilidade de ocorrer uma determinada quantidade de eventos em um intervalo de tempo, espaço ou quantidade, considerando uma taxa média de ocorrência. Na indústria de alimentos, pode ser usado para analisar o número de embalagens defeituosas encontradas em um lote de 1.000 unidades.

Assim, de forma resumida: o modelo de Bernoulli analisa uma única tentativa, o Binomial analisa várias tentativas com um número fixo de experimentos e o Poisson analisa a quantidade de ocorrências de um evento em determinado intervalo.
