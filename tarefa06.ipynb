**Regressão linear simples** é um método estatístico utilizado para analisar e modelar a relação entre duas variáveis quantitativas. Por meio dessa técnica, é possível estimar o valor de uma variável Y a partir do valor conhecido de uma variável X, utilizando uma reta. A equação da reta é dada por Y=a+bX, em que "a" representa o valor esperado de Y quando X=0, e "b" representa a variação esperada em Y para cada aumento de uma unidade em X. Prever valores e identificar tendências são exemplos de aplicações da regressão linear simples, que pode ser representada graficamente por meio de um gráfico de dispersão acompanhado da reta de regressão.

```python
import statsmodels.api as sm

# Define the independent variable (X) and dependent variable (Y)
X = df['Quantidade de lipídeos em 100g do corte cru sem gordura']
Y = df['Valor energético em 100g do produto cru (em Kcal)']

# Add a constant to the independent variable for the intercept term
X = sm.add_constant(X)

# Create and fit the OLS (Ordinary Least Squares) regression model
model = sm.OLS(Y, X)
results = model.fit()

# Print the summary of the regression results to see the coefficients
print(results.summary())
```

Com base na estimação do Python, temos que o termo "a" (=98,81 Kcal) representa a densidade calórica de uma carne que possui 0g de lipídeo. Logo, o termo "b" (=7,65 Kcal/g) nos diz que a cada 1g de lipídeo, ao valor energético adiciona-se 7,65 Kcal.

A **correlação de Pearson** é simplesmente uma nota de -1 a +1 que mostra o quanto duas coisas caminham juntas. Ela responde a uma pergunta simples: "Se a primeira coisa aumentar, o que acontece com a segunda?". A nota +1 significa que as duas variáveis são proporcionais, isso quer dizer que quando uma aumenta, a outra também aumenta. A nota 0 significa que as varáveis não tem relação entre elas, ou seja, o aumento ou a diminuição de uma não afeta a outra. Já a nota -1 siginifica que as duas varáveis são inversamente proporcionais, ou seja, quando uma aumenta, a outra diminui.

```python
lipideos = df['Quantidade de lipídeos em 100g do corte cru sem gordura']
valor_energetico = df['Valor energético em 100g do produto cru (em Kcal)']

correlacao_pearson = lipideos.corr(valor_energetico)

print(f'A Correlação de Pearson entre a quantidade de lipídeos e o valor energético é: {correlacao_pearson:.2f}')
```

Conforme calculado, a correlação linear de Pearson é 0,84. Isso quer dizer que as duas varáveis em questão (valor energético e quantidade de lipídeos) são proporcionais, entretanto, como não chegou a 1, podemos afirmar que há valores que fogem à regra.  

É importante salientar que a correlação linear de Pearson apenas calcula a relação entre duas variáveis com base numa reta. Isso quer dizer que, se duas variáveis apresentarem correlações mais complexas como uma curva no gráfico, esta correlação não as abrange. Por isso, o valor 0 calculado nem sempre significa que não há qualquer relação entre os dois dados variáveis.

## Fórmula da Regressão Linear Simples

A equação da reta de regressão linear simples é dada por:

$$Y = a + bX$$

## Fórmula da Correlação de Pearson

O coeficiente de correlação de Pearson ($\rho$) entre duas variáveis $X$ e $Y$ é calculado pela fórmula:

$$\rho(X, Y) = \frac{\sum_{i=1}^{n}(X_i - \bar{X})(Y_i - \bar{Y})}{\sqrt{\sum_{i=1}^{n}(X_i - \bar{X})^2 \sum_{i=1}^{n}(Y_i - \bar{Y})^2}}$$
