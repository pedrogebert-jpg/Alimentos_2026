1) **Distribuição normal de X**

Parâmetros: média 40; desvio padrão 4

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros da distribuição
mu = 40
sigma = 4

# Intervalo de valores para o eixo X (3 desvios-padrão para cada lado da média)
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, mu, sigma)

# Criação do gráfico
plt.figure(figsize=(10, 6))
plt.plot(x, y, label=rf'Normal($μ={mu}$, $σ={sigma}$)', color='blue', linewidth=2)

# Destacar a média no gráfico
plt.axvline(mu, color='red', linestyle='--', label=rf'Média ($μ$) = {mu}')

# Destacar as faixas de desvio-padrão
plt.axvline(mu - sigma, color='orange', linestyle=':', label=rf'$μ ± σ$ ({mu-sigma} e {mu+sigma})')
plt.axvline(mu + sigma, color='orange', linestyle=':')

# Preencher a área sob a curva
plt.fill_between(x, y, alpha=0.1, color='blue')

# Títulos e labels
plt.title('Distribuição de Probabilidade Normal de X (Tempo de Processamento)', fontsize=14)
plt.xlabel('Tempo X (minutos)', fontsize=12)
plt.ylabel('Densidade de Probabilidade', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)

# Mostrar o gráfico
plt.show()
```

```python
from scipy.stats import norm

# Parâmetros da distribuição
mu = 40
sigma = 4
limite = 35

# Cálculo da área (probabilidade acumulada até 35 minutos)
area = norm.cdf(limite, mu, sigma)

print(f"A área sob a curva até {limite} minutos é: {area:.6f} (ou {area * 100:.4f}%)")
```

```python
from scipy.stats import norm

# Parâmetros da distribuição
mu = 40
sigma = 4
limite_superior = 45

# Cálculo da probabilidade de levar pelo menos 45 minutos: P(X >= 45) = 1 - P(X < 45)
probabilidade = 1 - norm.cdf(limite_superior, mu, sigma)

print(f"A probabilidade de levar pelo menos {limite_superior} minutos é: {probabilidade:.6f} (ou {probabilidade * 100:.4f}%)")
```

3) Já que os números 35 e 45 são simétricos pelo fato da mpedia ser 40, podemos dizer que a área contemplada após os 45 minutos é igual a área até os 35 minutos, ou seja, 0,106 ou 10,6%

```python
from scipy.stats import norm

# Parâmetros da distribuição
mu = 40
sigma = 4
limite_inf = 35
limite_sup = 45

# Cálculo de P(35 <= X <= 45) = P(X <= 45) - P(X <= 35)
probabilidade_intervalo = norm.cdf(limite_sup, mu, sigma) - norm.cdf(limite_inf, mu, sigma)

print(f"A probabilidade de levar entre {limite_inf} e {limite_sup} minutos é: {probabilidade_intervalo:.6f} (ou {probabilidade_intervalo * 100:.4f}%)")
```

### 5) Gráficos das Áreas Calculadas sob a Distribuição Normal

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parâmetros da distribuição
mu = 40
sigma = 4

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = norm.pdf(x, mu, sigma)

fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

# --- Gráfico 1: Até 35 minutos ---
axes[0].plot(x, y, color='blue', linewidth=2)
x_fill1 = np.linspace(mu - 4*sigma, 35, 500)
y_fill1 = norm.pdf(x_fill1, mu, sigma)
axes[0].fill_between(x_fill1, y_fill1, color='red', alpha=0.4, label='Área < 35 min\n(10.57%)')
axes[0].axvline(35, color='red', linestyle='--')
axes[0].set_title('Área até 35 minutos', fontsize=12)
axes[0].set_xlabel('Tempo X (minutos)')
axes[0].set_ylabel('Densidade de Probabilidade')
axes[0].legend(loc='upper right')
axes[0].grid(True, alpha=0.3)

# --- Gráfico 2: Após 45 minutos ---
axes[1].plot(x, y, color='blue', linewidth=2)
x_fill2 = np.linspace(45, mu + 4*sigma, 500)
y_fill2 = norm.pdf(x_fill2, mu, sigma)
axes[1].fill_between(x_fill2, y_fill2, color='orange', alpha=0.4, label='Área > 45 min\n(10.57%)')
axes[1].axvline(45, color='orange', linestyle='--')
axes[1].set_title('Área após 45 minutos', fontsize=12)
axes[1].set_xlabel('Tempo X (minutos)')
axes[1].legend(loc='upper right')
axes[1].grid(True, alpha=0.3)

# --- Gráfico 3: Entre 35 e 45 minutos ---
axes[2].plot(x, y, color='blue', linewidth=2)
x_fill3 = np.linspace(35, 45, 500)
y_fill3 = norm.pdf(x_fill3, mu, sigma)
axes[2].fill_between(x_fill3, y_fill3, color='green', alpha=0.4, label='Área 35-45 min\n(78.87%)')
axes[2].axvline(35, color='green', linestyle='--')
axes[2].axvline(45, color='green', linestyle='--')
axes[2].set_title('Área entre 35 e 45 minutos', fontsize=12)
axes[2].set_xlabel('Tempo X (minutos)')
axes[2].legend(loc='upper right')
axes[2].grid(True, alpha=0.3)

plt.suptitle('Destaque das Áreas Calculadas sob a Distribuição Normal', fontsize=16, y=1.05)
plt.tight_layout()
plt.show()
```

### 6) Percentil de 95%

O percentil de 95% indica o tempo limite dentro do qual 95% de todos os processos de alimentos são concluídos. Para encontrar esse valor sob uma distribuição normal, utilizamos a função inversa da distribuição acumulada (conhecida como Percent Point Function ou PPF).

```python
from scipy.stats import norm

# Parâmetros da distribuição
mu = 40
sigma = 4
percentil = 0.95

# Cálculo do tempo limite correspondente a 95% de probabilidade acumulada
tempo_limite = norm.ppf(percentil, mu, sigma)

print(f"O percentil de 95% (P95) é de aproximadamente: {tempo_limite:.2f} minutos.")
print(f"Isso significa que 95% dos processos de alimentos duram até {tempo_limite:.2f} minutos.")
```

A resposta exata obtida é **46,58 minutos**.

Isso significa que, sob essa distribuição teórica ($
\mu = 40$
 e $
\sigma = 4$
), **95% dos processos de processamento de alimentos são concluídos em até 46,58 minutos** (restantes apenas 5% dos processos que demoram mais do que este período).

### 7) Tempo limite para 99% dos processamentos

Para estabelecer um tempo limite onde 99% de todos os processos sejam concluídos, calculamos o percentil de 99% (P99) da nossa distribuição normal ($
\mu = 40$
, $
\sigma = 4$
).

```python
from scipy.stats import norm

# Parâmetros da distribuição
mu = 40
sigma = 4
percentil_99 = 0.99

# Cálculo do tempo correspondente ao percentil de 99%
tempo_limite_99 = norm.ppf(percentil_99, mu, sigma)

print(f"O tempo limite para que 99% dos processos sejam concluídos é de aproximadamente: {tempo_limite_99:.2f} minutos.")
```

A resposta obtida é **49,31 minutos**.

Isso significa que, para garantir que 99% dos processamentos de alimentos sejam concluídos dentro do prazo estabelecido, o responsável deve definir um tempo limite de aproximadamente **49,31 minutos** (ou cerca de 49 minutos e 18 segundos).

### 8) Interpretação e Comparação dos Valores (Itens 6 e 7)

Ao analisarmos e compararmos os tempos limites obtidos nos itens anteriores, temos:

* **Percentil de 95% (Item 6):** **46,58 minutos**
* **Percentil de 99% (Item 7):** **49,31 minutos**

#### **Análise Comparativa e Interpretação:**

1. **Aumento Margem de Segurança:** Para elevar a garantia de conclusão dos processos de **95% para 99%** (um aumento de 4% na cobertura de segurança), o tempo limite precisa ser estendido em apenas **2,73 minutos** (de 46,58 para 49,31 minutos).
2. **Comportamento das Caudas da Distribuição Normal:** Esse acréscimo relativamente pequeno de tempo para uma cobertura de quase 100% ocorre porque, em uma curva normal, a maior parte dos dados está concentrada em torno da média ($
\mu = 40$
). Conforme nos afastamos rumo às extremidades (caudas), a densidade de probabilidade diminui rapidamente.
3. **Tomada de Decisão Prática:**
   * Se o responsável pela operação for mais tolerante a pequenos atrasos eventuais (5% de risco), pode adotar a meta de **46,58 minutos**.
   * Se o processo exigir extrema rigidez e quase nenhum atraso for permitido (apenas 1% de risco), o padrão ideal a ser estabelecido é de **49,31 minutos**.
