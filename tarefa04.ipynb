Conforme a legenda, a linha horizontal inferior do retângulo azul representa o **Primeiro Quartil (Q1)**, com um valor de 9,9 mil toneladas. A linha logo acima dele, dividindo a caixa, é a **Mediana (Q2)**, ou o segundo quartil. Deste modo, a linha acima, fechando o retângulo, é o **Terceiro Quartil (Q3)**. **Como determinamos se há observações atípicas (outliers) em uma base de dados?** Existem alguma estratégias para identificar essas observações. São cinco:

1.  **Método interquartil (IQR):** calculando os quartis, valores que ficam abaixo do Q1 e acima do Q3 podem ser valores atípicos.
2.  **Boxplot:** como já feito, o gráfico de Boxplot fornece informações visuais dos dados analisados. Os valores fora do retângulo central, são potenciais outliers.
3.  **Z-score:** este método mostra quantos desvios-padrão estão distantes da média. Para obtê-lo é necessário subtrair a média dos dados do valor que queremos analisar e dividir pelo desvio padrão. O valor encontrado é chamada, então de Z-score. Para z=0, o valor é exatamente igual à média. Se o valor de z=1, temos 1 desvio padrão acima da média, mas quando z=-1, sabe-se que também temos 1 desvio padrão, porém este está abaixo da média.
4.  **Distância de Mahalanobis:** este método analisa várias variáveis simultaneamente. A principal diferença deste método para o z-score, é que o z-score analisa somente uma varável por vez.

**Conceitos teóricos da estatística**

1) Mediana: Este conceito é amplamente utilizado quando se precisa de um número que represente o centro da distribuição de dados. Por exemplo: Se catalogássemos os salários de uma empresa X em que duas pessoas recebem um salário acima de R$40.000, enquanto o restante recebe salários abaixo dos cinco mil reais, fazendo uma média aritmética, o valor seria muito alto, enquanto isso não representaria a realidade.
Se colocarmos valores em ordem crescente numa linha e escolhermos o valor que está exatamente na metade (quanto se tem uma quantidade ímpar de valores), este representa a **mediana**. Se tivermos dois valores centrais (quanto temos um número par de dados), a mediana é a média aritmética desses dois valores.

2) Quartis: Utilizando do mesmo raciocínio para obter a mediana, após a ordenação crescente dos valores, obtem-se 3 valores de modo a dividi-los em quatro parte iguais. O primeiro valor obtido é denominado Q1 (primeiro quartil), o segundo se chama Q2 (segundo quartil, ou a mediana) e o terceiro é o Q3 (terceiro quartil). Utilizando desta técnica, podemos identificar, além da mediana, valores importantes de destaque, como os que são muito altos ou muito baixos.

3) Amplitude interquartílica: Podemos dizer que a amplitude quartílica é representada por ΔQ, ou seja, Q3-Q1. Esta análise é feita afim de observar se os valores do arranjo de dados estão concentrados ou mais dispersos em sua parte central, evitando distorções por valores muito altos ou muito baixos.

4) Boxplot: O boxplot é um tipo de gráfico utilizado para observar a distribuição de variáveis numéricas ordinais, utilizando os conceitos de mínimo, máximo, quartis e mediana. Com este formato de gráfico, também podemos visualizar com mais clareza os dados discrepantes, também conhecidos como *Outliers*. O gráfico possui uma linha vertical, onde a parte inferior mostra o valor mínimo e a parte superior mostra o valor máximo. Nessa linha temos um retângulo. A linha horizontal inferior no retângulo representa o primeiro quartil (Q1). Dividindo a figura em dois segmentos, está a linha horizontal que representa a mediana, ou seja, o Q2. Acima, na parte superior do polígono, aparece uma linha horizontal representando, então, o terceiro quartil (Q3).

5) *Outliers* ou observações atípicas: os dados numéricos que distoam da maioria dos outros valores. São dados "fora do comum" e que facilmente podem ser identificados.
